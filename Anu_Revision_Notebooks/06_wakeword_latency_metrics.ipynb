# 06 — Wake-word and latency metrics
This summarizes measurements from real continuous wake-word sessions and latency tests. It does not record microphone audio itself.

```python
from pathlib import Path
import pandas as pd
OUT=Path("anu_revision_output"); OUT.mkdir(exist_ok=True)
template=pd.DataFrame(columns=["session_id","network_condition","recording_hours","wakeword_trials",
 "correct_detections","false_activations","missed_detections","avg_latency_ms"])
template.to_csv(OUT/"wakeword_latency_sessions_TO_FILL.csv",index=False)
print("Fill wakeword_latency_sessions_TO_FILL.csv with actual measured sessions, then run the next cell.")
```

```python
df=pd.read_csv(OUT/"wakeword_latency_sessions_TO_FILLL.csv")
assert len(df)>0,"Enter real measurements first."
cols=["recording_hours","wakeword_trials","correct_detections","false_activations","missed_detections","avg_latency_ms"]
for c in cols: df[c]=pd.to_numeric(df[c],errors="coerce")
assert df[cols].notna().all().all(),"Complete all numeric columns."
assert (df.recording_hours>0).all()
trials=df.wakeword_trials.sum(); hours=df.recording_hours.sum()
print("Detection rate:",round(df.correct_detections.sum()/trials,4) if trials else "No trials")
print("Miss rate:",round(df.missed_detections.sum()/trials,4) if trials else "No trials")
print("False activations/hour:",round(df.false_activations.sum()/hours,4))
display(df.groupby("network_condition").avg_latency_ms.agg(["count","mean","median"]))
df.to_csv(OUT/"wakeword_latency_sessions.csv",index=False)
```
