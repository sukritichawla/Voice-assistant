# 01 — Dataset audit and manifest
Set DATASET_DIR to the folder containing one subfolder per command/class. This inventories WAVs and creates a manifest. You must fill speaker IDs and original-file relationships; the notebook will not guess them.

```python
from pathlib import Path
import pandas as pd

DATASET_DIR = Path(r"C:\Users\ADMIN\Desktop\Sukriti\VIT\SL\AnuAssistant\dataset")
OUT = Path("anu_revision_output")
OUT.mkdir(exist_ok=True)
assert DATASET_DIR.exists(), f"Dataset folder not found: {DATASET_DIR}"
```

```python
rows=[]
for folder in sorted(p for p in DATASET_DIR.iterdir() if p.is_dir()):
    for wav in sorted(folder.rglob("*.wav")):
        rows.append({"path":str(wav.resolve()),"label":folder.name,"filename":wav.name,
                     "speaker_id":"","original_file":"","is_augmented":""})
manifest=pd.DataFrame(rows)
print("WAV files:",len(manifest),"Classes:",manifest.label.nunique())
display(manifest.groupby("label").size().to_frame("count"))
manifest.to_csv(OUT/"dataset_manifest_TO_FILL.csv",index=False)
display(manifest.head())
```
