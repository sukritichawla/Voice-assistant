# 05 — End-to-end intent/entity/action metrics
Run real commands through the assistant, record the outcomes in the generated CSV, then calculate the metrics. No success is simulated.

```python
from pathlib import Path
import pandas as pd
OUT=Path("anu_revision_output"); OUT.mkdir(exist_ok=True)
template=pd.DataFrame(columns=["true_intent","predicted_intent","true_entity","predicted_entity","entity_required","action_success"])
template.to_csv(OUT/"pipeline_test_results_TO_FILL.csv",index=False)
print("Created pipeline_test_results_TO_FILL.csv. Fill one row per actual command attempt, then run the next cell.")
```

```python
df=pd.read_csv(OUT/"pipeline_test_results_TO_FILL.csv").fillna("")
assert len(df)>0,"Enter actual test attempts first."
norm=lambda x:str(x).strip().lower()
intent_ok=df.apply(lambda r:norm(r.true_intent)==norm(r.predicted_intent),axis=1)
needs=df.entity_required.astype(str).str.lower().isin(["true","1","yes"])
entity_ok=df.apply(lambda r:norm(r.true_entity)==norm(r.predicted_entity),axis=1)
joint=intent_ok & (~needs | entity_ok)
action=df.action_success.astype(str).str.lower().isin(["true","1","yes"])
print("Intent accuracy:",round(intent_ok.mean(),4))
print("Entity accuracy:",round(entity_ok[needs].mean(),4) if needs.any() else "No entity-required cases")
print("Joint intent/entity accuracy:",round(joint.mean(),4))
print("Successful action rate:",round(action.mean(),4))
print("End-to-end success rate:",round((joint & action).mean(),4))
```
